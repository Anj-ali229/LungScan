```python
# Create the Space
api.create_repo(f"{USERNAME}/lungscan", repo_type="space", space_sdk="gradio", exist_ok=True)
print("Space created")

# Upload the 3 app files
import os

files = {
    "app.py":          "/home/jupyter-projectm10/Major Project/webapp/app.py",           # wherever you saved it
    "requirements.txt":"/home/jupyter-projectm10/Major Project/webapp/requirements.txt",
    "README.md":       "/home/jupyter-projectm10/Major Project/webapp/README.md",
}

for repo_name, local_path in files.items():
    api.upload_file(
        path_or_fileobj=local_path,
        path_in_repo=repo_name,
        repo_id=f"{USERNAME}/lungscan",
        repo_type="space"
    )
    print(f"Uploaded {repo_name}")
```
